SERIES - A Series is a one-dimensional labeled array

Series has - index and its value 

```python
#Series from lists
```

```python
import numpy as np 
import pandas as pd
```

```python
#string
countries = ["India", "Canada", "USA", "Iceland"]   #python list

pd.Series(countries)  #Series is a class inside pandas and we made an object of it. dtype = object (99% times its string)
```

```python
#numeric
runs = [45, 60, 22, 89, 101]
ed = pd.Series(runs)
ed.index
```

```python
#customized index
marks = [98,75,58,92,68]
subject = ['Cse','Maths', 'Physics', 'English', 'Chemistry']

pd.Series(marks,index = subject)
```

```python
#setting a name
Marks= pd.Series(marks,index = subject, name="Preetika's Marks")
Marks
```

```python
#Series from dictionaries
```

```python
marks = {'maths':67,
         'science':80,
         'hindi':45,
         'english':97}

marks_series = pd.Series(marks, name = 'preetika marks')
marks_series    #in dictionary by deafult key becomes index and pair becomes values
```

#SERIES INDEXING 

```python
#integer indexing
a = pd.Series([2,4,6,8,10,12])
a[2]
```

```python
#negative indexing - When your custom index is strings, 
#negative indexing will work just like in a Python list, 
#because strings don't conflict with positional indexing. 
#However, when the index is numeric, pandas assumes you’re 
#trying to use labels as well. Since it doesn’t find a label “-1” 
#(or any negative number) in a purely numeric index, it throws an error.

```

```python
#slicing
vk[5:16]
```

```python
#negative slicing
vk[-5:]
```

```python
#fancy indexing
vk[[1,2,4,5]]
```

```python
#indexing with labels -> fancy indexing
movies['Hum Tumhare Hain Sanam']
```
